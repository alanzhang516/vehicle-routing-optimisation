```python
import pandas as pd
import numpy as np
```

```python
demands = pd.read_csv('Data/demand.csv')
durations = pd.read_csv('Data/FoodstuffsDurations2026.csv')
demands_cleaned = demands.drop(columns=['Mon mean', 'Tue mean', 'Wed mean', 'Thu mean', 'Fri mean', 'Mon rounded', 'Tue rounded', 'Wed rounded', 'Thu rounded', 'Fri rounded', 'Sat rounded'])
print(demands_cleaned)
demands_cleaned.to_csv('Data/cleaned_demand.csv', index=False)
```

```python
deviation = pd.read_csv('Data/DemandDeviation.csv')
sd_cols = ["Mon SD", "Tue SD", "Wed SD", "Thu SD", "Fri SD"]
deviation_cleaned = deviation[["Supermarket","Sat SD"]].copy()
deviation_cleaned["Weekday SD"] = np.sqrt((deviation[sd_cols] ** 2).mean(axis=1))
deviation_cleaned.to_csv('deviation_cleaned.csv', index=False)
```
